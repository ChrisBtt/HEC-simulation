```python
import os
import numpy as np
import pandas as pd
import scipy as sp
import topas2numpy as tn
import matplotlib.pyplot as plt
from ipywidgets import interact
import matplotlib.colors as colors
import matplotlib.ticker as ticker

import seaborn as sns

import sys
sys.path.append('..')
from preprocessSimulation import Preprocessor
```

```python
root = 'H:/EIT/Christoph/Monte_Carlo/05_motor_servo_wood_air_cylinder_setupA'
# root = '../../TOPAS_simulation_data/motor_servo/'

print(os.listdir(root))
folders = [f for f in os.listdir(root) if os.path.isdir(os.path.join(root, f)) and '.' not in f and f.startswith('F7x1')]
print(folders)

data = {}
for folder in folders:
    path_dose = f"{root}/{folder}/DoseGrid.csv"
    path_current_elX = f"{root}/{folder}/velCurrElX.csv"
    path_current_elY = f"{root}/{folder}/velCurrElY.csv"
    path_current_elZ = f"{root}/{folder}/velCurrElZ.csv"

    bin_x = 0.5
    bin_y = 0.5
    bin_z = 0.1

    data[folder] = Preprocessor([path_dose, path_current_elX, path_current_elY, path_current_elZ], bin_x, bin_y, bin_z, center=True).data
```

```python

variance = [0.5, 15]
fig, ax = plt.subplots(1, 2, figsize=(12,6), constrained_layout=True)

data_filter = {}
# ax_y = []
for j, (folder, dat) in enumerate(data.items()):
    print(f"Analyzing data from folder: {folder}")
    print(f"Data columns: {dat.columns}")

    center_x = (dat['x_cm'].max() + dat['x_cm'].min()) / 2
    center_y = (dat['y_cm'].max() + dat['y_cm'].min()) / 2
    print(f'Center Fine Scoring at x={center_x} cm for x_list={dat["x_cm"].unique()}')
    for i, var in enumerate(variance):
        # ax_y.append(ax[i].twinx())
        print(f'Filtering with variance: {var} cm')
        filter_center = (dat['x_cm'].between(center_x - var, center_x + var)) & (dat['y_cm'].between(center_y - var, center_y + var))
        data_filtered = dat[filter_center]
        key = f"{folder}_var{var}"
        data_filter[key] = data_filtered
        print(f"Range x values after filtering: min={data_filtered['x_cm'].min()} cm, max={data_filtered['x_cm'].max()} cm")
        print(f"Range y values after filtering: min={data_filtered['y_cm'].min()} cm, max={data_filtered['y_cm'].max()} cm")

        if j==0:
            if i == 0:
                # norm_dose_1 = data_filtered.groupby('z_cm')['dose'].mean().max()
                norm_current_1 = data_filtered.groupby('z_cm')['jz_cm'].mean().max()
            else:
                # norm_dose_2 = data_filtered.groupby('z_cm')['dose'].mean().max()
                norm_current_2 = data_filtered.groupby('z_cm')['jz_cm'].mean().max()
            # plt2 = sns.lineplot(data=data_filtered, x='z_cm', y='dose', ax=ax[i], label=f'dose, {folder}', errorbar=None)
            plt2 = sns.lineplot(data=data_filtered, x='z_cm', y='jz_cm', ax=ax[i], label=f'jz, {folder}', errorbar=None)
        else:
            if i == 0:
                # scale_dose = data_filtered.groupby('z_cm')['dose'].mean().max() / norm_dose_1
                scale_current = data_filtered.groupby('z_cm')['jz_cm'].mean().min() / norm_current_1
                # data_filtered['dose_norm'] = data_filtered['dose'] / scale_dose
                data_filtered['jz_cm_norm'] = data_filtered['jz_cm'] / scale_current
            else:
                scale_current = data_filtered.groupby('z_cm')['jz_cm'].mean().min() / norm_current_2
                data_filtered['jz_cm_norm'] = data_filtered['jz_cm'] / scale_current
            
            # plt4 = sns.lineplot(data=data_filtered, x='z_cm', y='jz_cm_norm', ax=ax[i], label=f'jz, {folder}', errorbar=None)
            plt4 = sns.lineplot(data=data_filtered, x='z_cm', y='jz_cm_norm', ax=ax[i], label=f'jz, {folder}', errorbar=None)

        ax[i].set_xlabel('Depth (cm)')
        ax[i].set_ylabel('Current density (normalized)')
        # ax_y[i].set_ylabel('Current density (normalized)')
        title = 'Metrics in central voxel' if i == 0 else 'Metrics averaged over x-y-plane'
        ax[i].set_title(title)
        ax[i].legend(loc='lower right')
        ax[i].grid()
```

## Radial current density

```python
depth = 3

fig, ax = plt.subplots(2, 2, figsize=(12, 5), constrained_layout=True)
plt.suptitle(f'Current Profile at Depth = ({depth}+- 0.5) cm')

for i, (key, data_current_el) in enumerate(data.items()):

    filter_depth = (data_current_el['z_cm'] > depth-0.5) & (data_current_el['z_cm'] < depth+0.5)
    data_current_el_filtered = data_current_el[filter_depth]
    data_current_el_filtered_x = data_current_el_filtered[(data_current_el_filtered['y_cm'] == int(data_current_el_filtered['y_cm'].median()))]
    data_current_el_filtered_y = data_current_el_filtered[(data_current_el_filtered['x_cm'] == int(data_current_el_filtered['x_cm'].median()))]

    pivot_data_x = data_current_el_filtered.pivot_table(index='y_cm', columns='x_cm', values='jx_cm')
    pivot_data_y = data_current_el_filtered.pivot_table(index='y_cm', columns='x_cm', values='jy_cm')

    sns.lineplot(data=data_current_el_filtered_x, x='x_cm', y='jx_cm', ax=ax[0, 0], label=key, errorbar=None)
    ax[0, 0].set_xlabel('x (cm)')
    ax[0, 0].set_ylabel('Current (q/cm²)')
    ax[0, 0].grid()

    if i == 0:
        sns.heatmap(pivot_data_x, cmap='viridis', ax=ax[1, 0], cbar_kws={'label': 'X-Current (q/cm²)'})
        ax[1, 0].xaxis.set_major_formatter(ticker.FormatStrFormatter('%.d'))
        ax[1, 0].yaxis.set_major_formatter(ticker.FormatStrFormatter('%.d'))
        ax[1, 0].set_xlabel(r'X Position (cm)')
        ax[1, 0].set_ylabel(r'Y Position (cm)')

    sns.lineplot(data=data_current_el_filtered_y, x='y_cm', y='jy_cm', ax=ax[0, 1], label=key, errorbar=None)
    ax[0, 1].set_xlabel('y (cm)')
    ax[0, 1].set_ylabel('Current (q/cm²)')
    ax[0, 1].grid()

    if i == 0:
        sns.heatmap(pivot_data_y, cmap='viridis', ax=ax[1, 1], cbar_kws={'label': 'Y-Current (q/cm²)'})
        ax[1, 1].xaxis.set_major_formatter(ticker.FormatStrFormatter('%.d'))
        ax[1, 1].yaxis.set_major_formatter(ticker.FormatStrFormatter('%.d'))
        ax[1, 1].set_xlabel(r'X Position (cm)')
        ax[1, 1].set_ylabel(r'Y Position (cm)')
```
